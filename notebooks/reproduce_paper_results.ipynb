## 2. DDI calculation example

```python
justice_decomp = ['fairness', 'rights', 'obligations', 'punishment', 'compassion']
justice_locked = ['cultural_bias', 'emotional_charge']
ddi_justice = compute_ddi(justice_decomp, justice_locked)
print(f"DDI of 'justice' = {ddi_justice:.3f}")

# Lock dynamics
L = 0.6
for t in range(10):
    ddi = compute_ddi(justice_decomp, justice_locked)
    L = update_lock_degree(L, alpha=0.05, beta=0.12, ddi=ddi,
                           information_input=1.0, decomposition_activity=ddi)
    print(f"Step {t+1}: DDI={ddi:.3f}, Lock degree={L:.3f}")
```
